# Task 1: Exploratory Data Analysis (EDA)

**Objective:** Perform comprehensive EDA on a structured customer dataset using Pandas, Matplotlib and Seaborn.

**Dataset:** `customer_data.csv` (500 synthetic customer records created for this demonstration). Replace it with an organization-approved dataset if your internship requires a real company dataset.

**Workflow:** load data → inspect structure → summarize statistics → check data quality → visualize distributions and relationships → record findings.

```python
# Install if required (run once in your environment)
# %pip install pandas numpy matplotlib seaborn

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
df = pd.read_csv("customer_data.csv")
df.head()
```

## 1. Dataset structure and quality

```python
print("Shape:", df.shape)
print("\\nColumn information:")
df.info()
print("\\nDuplicate rows:", df.duplicated().sum())
print("\\nMissing values by column:")
print(df.isna().sum())
```

## 2. Descriptive statistics

```python
display(df.describe(include="all").T)
print("\\nNumeric column summaries:")
display(df.select_dtypes(include=np.number).describe().T)
```

## 3. Missing-value visualization

```python
missing_pct = (df.isna().mean() * 100).sort_values(ascending=False)
missing_pct = missing_pct[missing_pct > 0]
plt.figure(figsize=(9,4))
sns.barplot(x=missing_pct.values, y=missing_pct.index)
plt.xlabel("Missing values (%)")
plt.ylabel("Column")
plt.title("Percentage of missing values by column")
plt.tight_layout()
plt.show()
```

## 4. Univariate analysis

```python
numeric_cols = ["age", "annual_income", "spending_score", "visits_per_month"]
df[numeric_cols].hist(figsize=(11,8), bins=20, edgecolor="black")
plt.suptitle("Distributions of numerical customer attributes")
plt.tight_layout()
plt.show()

for col in ["gender", "membership", "region"]:
    plt.figure(figsize=(6,3))
    sns.countplot(data=df, x=col, order=df[col].value_counts().index)
    plt.title(f"Distribution of {col}")
    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()
```

## 5. Outlier inspection with boxplots

```python
fig, axes = plt.subplots(1, len(numeric_cols), figsize=(14,4))
for ax, col in zip(axes, numeric_cols):
    sns.boxplot(y=df[col], ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

## 6. Correlation analysis

```python
corr = df[numeric_cols].corr(numeric_only=True)
plt.figure(figsize=(7,5))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation heatmap")
plt.tight_layout()
plt.show()
```

## 7. Relationship analysis

```python
plt.figure(figsize=(7,5))
sns.scatterplot(data=df, x="annual_income", y="spending_score", hue="membership", alpha=.75)
plt.title("Annual income vs spending score")
plt.tight_layout()
plt.show()

plt.figure(figsize=(7,4))
sns.boxplot(data=df, x="membership", y="spending_score", order=["Basic","Silver","Gold"])
plt.title("Spending score by membership")
plt.tight_layout()
plt.show()
```

## 8. Findings and conclusion

After running the notebook, record findings based on the actual output:

- Boxplots help flag unusually high income observations for investigation; an outlier should not be removed automatically.
